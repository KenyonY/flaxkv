```python
from flaxkv2 import FlaxKV
import pandas as pd
import numpy as np
```

```python
test_cases = [
         (5, "少于20个，应显示所有"),
         (20, "正好20个，应显示所有"),
         (25, "超过20个，应显示前20个"),
         (50, "50个，应显示前20个")
     ]

for num_items, description in test_cases:
    print(f"\n{'='*60}")
    print(f"测试: {num_items}个键值对 - {description}")
    print('='*60)
    
    db = FlaxKV(f'test_{num_items}', 'test_dbs', rebuild=True)
    
    # 添加数据
    for i in range(num_items):
        db[f'key_{i:03d}'] = i
    print(db)

 
```

```python
db.stat()
```

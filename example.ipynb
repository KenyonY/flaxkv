```python
from flaxkv2 import FlaxKV
from flaxkv2.client.zmq_client import RemoteDBDict
import pandas as pd
import numpy as np
from sparrow import MeasureTime
import random
mt = MeasureTime()

read_cache_size=512 * 1024 * 1024
write_buffer_size=512 * 1024 * 1024

# read_cache_size=512 * 1024 * 1024
# write_buffer_size=0

# read_cache_size=0
# write_buffer_size=512 * 1024 * 1024

db = FlaxKV("kunyuan", auto_nested=True, rebuild=True, 
            read_cache_size=read_cache_size,
            write_buffer_size=write_buffer_size
           ) 

db_no_nested = FlaxKV("kunyuan_no_nested", 
                      auto_nested=False, rebuild=True,
            read_cache_size=read_cache_size,
            write_buffer_size=write_buffer_size)

big_dict = {}
big_list = [random.random() for _ in range(1000)]
for i in range(100):
    big_dict[f"lst{i}"] = big_list

mt.start()
db['dict'] = big_dict
mt.show_interval("递归 写入")

db_no_nested['dict'] = big_dict
mt.show_interval("非递归 写入")

mt.start()
print(db['dict']['lst2'][2])
mt.show_interval("递归 读取")
print(db_no_nested['dict']['lst2'][2])
mt.show_interval("非递归 读取")
print()
```

```python
test_cases = [
         (5, "少于20个，应显示所有"),
         (20, "正好20个，应显示所有"),
         (25, "超过20个，应显示前20个"),
         (50, "50个，应显示前20个")
     ]

for num_items, description in test_cases:
    print(f"\n{'='*60}")
    print(f"测试: {num_items}个键值对 - {description}")
    print('='*60)
    
    db = FlaxKV(f'test_{num_items}', 'test_dbs', rebuild=True)
    
    # 添加数据
    for i in range(num_items):
        db[f'key_{i:03d}'] = i
    print(db)

 
```

```python
from flaxkv2 import RawLevelDBDict, CachedLevelDBDict
import numpy as np
from sparrow import MeasureTime
```

```python
from flaxkv import FlaxKV
```

```python

mt = MeasureTime()


# 生成测试数据集：2000条1000维的随机numpy数组
n = 10000
dim = 1000
test_data = np.random.rand(n, dim)

db1 = FlaxKV("flaxkv_1", rebuild=True)
db2 = RawLevelDBDict("flaxkv_2", rebuild=True,enable_ttl_cleanup=False)
db2_cached = CachedLevelDBDict("flaxkv_2_cached", read_cache_size=n, performance_profile='read_optimized', enable_ttl_cleanup=False, rebuild=True)

mt.start()
for i in range(n):
    db1[f"key_{i}"] = test_data[i]
mt.show_interval()

for i in range(n):
    db2[f"key_{i}"] = test_data[i]
mt.show_interval()


for i in range(n):
    db2_cached[f"key_{i}"] = test_data[i]
mt.show_interval()
```
